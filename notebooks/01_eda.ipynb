### 3. Key Observations & Takeaways
1. **Severe Imbalance**: 98.89% legitimate vs 1.11% fraudulent transactions (~1:89 ratio).
2. **Amount Skew**: Fraudulent transactions have a significantly higher mean (.44 vs .72) and a long tail up to .50 due to account takeover (Scenario 3) and high amount (Scenario 1) fraud.
3. **Log Transformation**: Applying log(1 + TX_AMOUNT) stabilizes the variance and compresses the heavy right-tail.
4. **Temporal Consistency**: Daily fraud volume stays consistent, requiring chronological splits rather than random shuffling to prevent future information leakage.
